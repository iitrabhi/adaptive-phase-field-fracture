```python
import gmsh

gmsh.initialize()

gmsh.model.add("glacier_terminus")

# -------------------------------------------------------------------------
# Geometry dimensions [m]
# -------------------------------------------------------------------------

Lx = 500.0
Ly = 750.0
Lz = 125.0

# -------------------------------------------------------------------------
# Notch dimensions [m]
# -------------------------------------------------------------------------

lx = 5.0
ly = 10.0
lz = 10.0

# Spacing between notch centers [m]
s = 50.0

# -------------------------------------------------------------------------
# Mesh size controls [m]
# -------------------------------------------------------------------------

mesh_factor = 1.0

h_min_base = 2.0
h_max_base = 18

h_min = mesh_factor * h_min_base
h_max = mesh_factor * h_max_base

gmsh.option.setNumber("Mesh.MeshSizeMin", h_min)
gmsh.option.setNumber("Mesh.MeshSizeMax", h_max)
gmsh.option.setNumber("Mesh.MeshSizeFactor", 1.0)

# -------------------------------------------------------------------------
# Glacier geometry
# -------------------------------------------------------------------------

glacier = gmsh.model.occ.addBox(
    0.0,
    0.0,
    0.0,
    Lx,
    Ly,
    Lz,
)

# -------------------------------------------------------------------------
# Five equally spaced notches on y = 0
#
# Notch center positions:
#
# x = Lx/2 - 2s
# x = Lx/2 - s
# x = Lx/2
# x = Lx/2 + s
# x = Lx/2 + 2s
# -------------------------------------------------------------------------

x_center = 0.5 * Lx

notch_centers = [
    x_center - 2.0 * s,
    x_center - 1.0 * s,
    x_center,
    x_center + 1.0 * s,
    x_center + 2.0 * s,
]

notches = []

for xc in notch_centers:

    notch_x0 = xc - 0.5 * lx
    notch_y0 = 0.0
    notch_z0 = Lz - lz

    notch = gmsh.model.occ.addBox(
        notch_x0,
        notch_y0,
        notch_z0,
        lx,
        ly,
        lz,
    )

    notches.append((3, notch))

# -------------------------------------------------------------------------
# Subtract all five notches
# -------------------------------------------------------------------------

domain, _ = gmsh.model.occ.cut(
    [(3, glacier)],
    notches,
    removeObject=True,
    removeTool=True,
)

gmsh.model.occ.synchronize()

# -------------------------------------------------------------------------
# Physical volume
# -------------------------------------------------------------------------

volume_tags = [tag for dim, tag in domain if dim == 3]

gmsh.model.addPhysicalGroup(3, volume_tags, 1)
gmsh.model.setPhysicalName(3, 1, "GLACIER")

# -------------------------------------------------------------------------
# Generate tetrahedral mesh
# -------------------------------------------------------------------------

gmsh.model.mesh.generate(3)

gmsh.write("glacier_terminus_5_notches_50.msh")

gmsh.finalize()
```

```python
import gmsh

gmsh.initialize()

gmsh.model.add("glacier_terminus")

# -------------------------------------------------------------------------
# Geometry dimensions [m]
# -------------------------------------------------------------------------

Lx = 500.0
Ly = 750.0
Lz = 125.0

# -------------------------------------------------------------------------
# Notch dimensions [m]
# -------------------------------------------------------------------------

lx = 5.0
ly = 10.0
lz = 10.0

# Spacing between notch centers [m]
s = 70.0

# -------------------------------------------------------------------------
# Mesh size controls [m]
# -------------------------------------------------------------------------

mesh_factor = 1.0

h_min_base = 2.0
h_max_base = 17.423

h_min = mesh_factor * h_min_base
h_max = mesh_factor * h_max_base

gmsh.option.setNumber("Mesh.MeshSizeMin", h_min)
gmsh.option.setNumber("Mesh.MeshSizeMax", h_max)
gmsh.option.setNumber("Mesh.MeshSizeFactor", 1.0)

# -------------------------------------------------------------------------
# Glacier geometry
# -------------------------------------------------------------------------

glacier = gmsh.model.occ.addBox(
    0.0,
    0.0,
    0.0,
    Lx,
    Ly,
    Lz,
)

# -------------------------------------------------------------------------
# Five equally spaced notch locations
#
# Centers:
#   Lx/2 - 2s
#   Lx/2 - s
#   Lx/2
#   Lx/2 + s
#   Lx/2 + 2s
#
# The same five notches are placed on:
#   y = 0
#   y = Ly
# -------------------------------------------------------------------------

x_center = 0.5 * Lx

notch_centers = [
    x_center - 2.0 * s,
    x_center - 1.0 * s,
    x_center,
    x_center + 1.0 * s,
    x_center + 2.0 * s,
]

notches = []

for xc in notch_centers:

    # ---------------------------------------------------------------------
    # Notch on y = 0 side
    # ---------------------------------------------------------------------

    notch_bottom = gmsh.model.occ.addBox(
        xc - 0.5 * lx,
        0.0,
        Lz - lz,
        lx,
        ly,
        lz,
    )

    notches.append((3, notch_bottom))

    # ---------------------------------------------------------------------
    # Opposite notch on y = Ly side
    # ---------------------------------------------------------------------

    notch_top = gmsh.model.occ.addBox(
        xc - 0.5 * lx,
        Ly - ly,
        Lz - lz,
        lx,
        ly,
        lz,
    )

    notches.append((3, notch_top))

# -------------------------------------------------------------------------
# Subtract all 10 notches
# -------------------------------------------------------------------------

domain, _ = gmsh.model.occ.cut(
    [(3, glacier)],
    notches,
    removeObject=True,
    removeTool=True,
)

gmsh.model.occ.synchronize()

# -------------------------------------------------------------------------
# Physical volume
# -------------------------------------------------------------------------

volume_tags = [tag for dim, tag in domain if dim == 3]

gmsh.model.addPhysicalGroup(3, volume_tags, 1)
gmsh.model.setPhysicalName(3, 1, "GLACIER")

# -------------------------------------------------------------------------
# Generate tetrahedral mesh
# -------------------------------------------------------------------------

gmsh.model.mesh.generate(3)

gmsh.write("glacier_terminus_10_notches_70.msh")

gmsh.finalize()
```

```python
import meshio
import os

os.makedirs("adaptive", exist_ok=True)

for file in os.listdir("."):
    if file.endswith(".msh"):
        mesh = meshio.read(file)
        cells = mesh.get_cells_type("tetra")
        points = mesh.points
        out_file = os.path.splitext(file)[0] + ".xdmf"
        meshio.write(
            os.path.join("adaptive", out_file),
            meshio.Mesh(points=points, cells={"tetra": cells}),
        )
```
