```python
import meshio
```

```python
mesh = meshio.read("Mesh_R.xdmf")
```

```python
cells = mesh.get_cells_type("tetra")
points = mesh.points

points_old = points.copy()
points[:, 0] = points_old[:, 1]
points[:, 1] = points_old[:, 0] 

meshio.write("Lx500_10C_S70.xdmf", meshio.Mesh(
    points=points,
    cells={"tetra": cells}))
```
