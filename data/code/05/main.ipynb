```python
import meshio
```

```python
mesh = meshio.read("tetra_S100.xdmf")
```

```python
cells = mesh.get_cells_type("tetra")
points = mesh.points

points_old = points.copy()
points[:, 0] = points_old[:, 1]
points[:, 1] = points_old[:, 0] 

meshio.write("Lx500_2C_S50.xdmf", meshio.Mesh(
    points=points,
    cells={"tetra": cells}))
```

```python
mesh = meshio.read("Lx500_2C_S40.xdmf")
cells = mesh.get_cells_type("tetra")
points = mesh.points
meshio.write("Lx500_2C_S50.xdmf", meshio.Mesh(
    points=points,
    cells={"tetra": cells}))
```
